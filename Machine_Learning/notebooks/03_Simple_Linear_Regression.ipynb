# Simple Linear Regression

One feature predicts a continuous target through a straight line. We use years of experience to predict salary.

## Imports and loading helper

```python
# Standard imports for the whole course
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Find the Datasets folder whether the notebook is run from the
# project root or from inside the notebooks folder.
from pathlib import Path
for candidate in [Path("Datasets"), Path("../Datasets")]:
    if candidate.exists():
        DATA = candidate
        break

def load(name):
    """Load a CSV from the Datasets folder by file name."""
    return pd.read_csv(DATA / name)

print("Datasets are in:", DATA.resolve())
print(sorted(p.name for p in DATA.glob("*.csv")))
```

## Loading and splitting the data

```python
df = load("Linear_Regression_Salary_Data.csv")
print(df.head())

X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0)
```

## Fitting the model

`fit` learns the intercept and slope that minimise the squared error on the training rows.

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
print("Intercept:", regressor.intercept_)
print("Slope:", regressor.coef_[0])
```

## Visualising the fit on the training set

```python
plt.scatter(X_train, y_train, color="green")
plt.plot(X_train, regressor.predict(X_train), color="red")
plt.title("Salary against experience (training set)")
plt.xlabel("Years of experience"); plt.ylabel("Salary"); plt.show()
```

## Visualising the fit on the test set

The test points were never seen during training, so a good fit here means the model generalises.

```python
plt.scatter(X_test, y_test, color="blue")
plt.plot(X_test, regressor.predict(X_test), color="red")
plt.title("Salary against experience (test set)")
plt.xlabel("Years of experience"); plt.ylabel("Salary"); plt.show()
```

## Scoring the model

The root mean squared error and R squared tell us how close the predictions are.

```python
from sklearn import metrics
y_pred = regressor.predict(X_test)
print("Mean salary:", df["Salary"].mean())
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print("R2 (%):", regressor.score(X_test, y_test) * 100)
```

## Predicting a new value

A single prediction still needs a two dimensional input, hence the double brackets.

```python
print(regressor.predict([[3.5]]))
```
