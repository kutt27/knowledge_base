# Naive Bayes, Decision Trees, and Random Forests

Three classification families: a probabilistic model, a single question based tree, and an ensemble of trees that vote.

## Imports and loading helper

```python
# Standard imports for the whole course
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Find the Datasets folder whether the notebook is run from the
# project root or from inside the notebooks folder.
from pathlib import Path
for candidate in [Path("Datasets"), Path("../Datasets")]:
    if candidate.exists():
        DATA = candidate
        break

def load(name):
    """Load a CSV from the Datasets folder by file name."""
    return pd.read_csv(DATA / name)

print("Datasets are in:", DATA.resolve())
print(sorted(p.name for p in DATA.glob("*.csv")))
```

## Preparing the customer data

```python
df = load("Naive_Bayes_Decision_RF_Tree_Boosting_User_Data.csv")
X = df.iloc[:, [2, 3]].values
y = df.iloc[:, 4].values

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Naive Bayes

The Gaussian variant assumes each feature follows a normal distribution.

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

nb = GaussianNB().fit(X_train, y_train)
print("Naive Bayes accuracy:", accuracy_score(y_test, nb.predict(X_test)))
```

## Decision Tree

The tree splits the data on the feature that best reduces entropy at each node.

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree

dt = DecisionTreeClassifier(criterion="entropy", random_state=0)
dt.fit(X_train, y_train)
print("Decision tree accuracy:", accuracy_score(y_test, dt.predict(X_test)))
```

## A small tree we can actually read

Limiting the depth keeps the tree small enough to picture.

```python
small = DecisionTreeClassifier(criterion="entropy", random_state=0,
                               max_depth=3).fit(X_train, y_train)
plt.figure(figsize=(11, 5))
plot_tree(small, feature_names=["Age", "EstimatedSalary"],
          class_names=["Not purchased", "Purchased"],
          filled=True, rounded=True, fontsize=8)
plt.show()
```

## Random Forest

Many trees, each on a random sample of rows and features, vote for the final answer.

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=10, criterion="entropy",
                            random_state=0)
rf.fit(X_train, y_train)
print("Random forest accuracy:", accuracy_score(y_test, rf.predict(X_test)))
```

A forest is steadier than a lone tree because the trees make different mistakes that tend to cancel.
