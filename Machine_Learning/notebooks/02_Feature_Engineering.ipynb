# Feature Engineering

Feature engineering uses knowledge of the problem to build better input columns. Here we practise the standard transformations: scaling, logarithms, binning, date extraction, and encoding.

## Creating a feature from two others

A ratio is often more informative than either raw column. The price to earnings ratio is the classic financial example.

```python
fin = pd.DataFrame({"Stock_Price": [100, 120, 140, 160, 180],
                      "Earnings_Per_Share": [5, 6, 7, 8, 9]})
fin["PE_Ratio"] = fin["Stock_Price"] / fin["Earnings_Per_Share"]
print(fin)
```

## Log transform for skewed data

A logarithm compresses large values and spreads out small ones.

```python
v = pd.Series([1, 4, 9, 16, 25], dtype=float)
plt.plot(v, np.log(v), "o-")
plt.xlabel("Original value"); plt.ylabel("Natural log"); plt.show()
```

## Binning a continuous column

Grouping age into bands turns a continuous feature into an ordered categorical one.

```python
ages = pd.DataFrame({"Age": [12, 35, 45, 23, 67, 89]})
ages["Age_Group"] = pd.cut(ages["Age"], bins=[0, 18, 35, 55, 100],
                           labels=["0-18", "19-35", "36-55", "56+"])
print(ages)
```

## Extracting features from a date

A single timestamp can be replaced by year, month, day, hour, and weekday, which a model can use directly.

```python
d = pd.DataFrame({"timestamp": ["2023-01-01 12:00:00",
                                    "2023-02-01 14:30:00",
                                    "2023-03-01 16:45:00"]})
d["timestamp"] = pd.to_datetime(d["timestamp"])
d["year"] = d["timestamp"].dt.year
d["month"] = d["timestamp"].dt.month
d["day"] = d["timestamp"].dt.day
d["weekday"] = d["timestamp"].dt.weekday
print(d)
```

## Encoding categorical features

One hot encoding is the safe default for a feature with no natural order.

```python
cat = pd.DataFrame({"Category": ["A", "B", "A", "C", "B"]})
print(pd.get_dummies(cat, columns=["Category"]))
```

These transformations are the ones you will reuse whenever a model performs poorly on the raw columns.
