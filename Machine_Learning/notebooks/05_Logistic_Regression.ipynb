# Logistic Regression

Logistic regression predicts a category by first predicting a probability with the sigmoid function, then applying a threshold.

## Imports and loading helper

```python
# Standard imports for the whole course
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

# Find the Datasets folder whether the notebook is run from the
# project root or from inside the notebooks folder.
from pathlib import Path
for candidate in [Path("Datasets"), Path("../Datasets")]:
    if candidate.exists():
        DATA = candidate
        break

def load(name):
    """Load a CSV from the Datasets folder by file name."""
    return pd.read_csv(DATA / name)

print("Datasets are in:", DATA.resolve())
print(sorted(p.name for p in DATA.glob("*.csv")))
```

## The sigmoid function

It squeezes any real number into the range zero to one.

```python
z = np.linspace(-8, 8, 200)
plt.plot(z, 1 / (1 + np.exp(-z)))
plt.axhline(0.5, color="gray", ls="--", lw=0.8)
plt.xlabel("z"); plt.ylabel("probability")
plt.title("The logistic (sigmoid) function"); plt.show()
```

## Loading the diabetes data

The file has no header, so we supply the column names.

```python
col_names = ["pregnant", "glucose", "bp", "skin", "insulin",
             "bmi", "pedigree", "age", "label"]
df = pd.read_csv(DATA / "Logistic_Regression_pima-indians-diabetes.csv",
                 header=None, names=col_names)
feature_cols = col_names[:-1]

X = df[feature_cols]
y = df["label"]

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0)
```

## Fitting the classifier and scoring it

```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression(solver="lbfgs", max_iter=1000)
logreg.fit(X_train, y_train)
y_pred = logreg.predict(X_test)

from sklearn import metrics
print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
print("Confusion matrix:\n", metrics.confusion_matrix(y_test, y_pred))
```

## Why glucose carries the signal

Plotting the glucose readings by outcome shows the two groups sit at different levels.

```python
plt.hist(df[df.label == 0]["glucose"], bins=20, alpha=0.6,
         label="not diabetic")
plt.hist(df[df.label == 1]["glucose"], bins=20, alpha=0.6,
         label="diabetic")
plt.xlabel("Blood glucose"); plt.ylabel("Count")
plt.legend(); plt.show()
```
